## Latihan 1 — Membuat Class Hero
**Tujuan:** Memahami cara membuat class dan atribut instance.

````python
class Hero:
    # Constructor: Dijalankan saat Hero baru dibuat
    def __init__(self, name, hp, attack_power):
        self.name         = name          # Nama Hero
        self.hp           = hp            # Nyawa (Health Point)
        self.attack_power = attack_power  # Kekuatan Serangan

    # Method untuk menampilkan info hero
    def info(self):
        print(f"Hero: {self.name} | HP: {self.hp} | Power: {self.attack_power}")

# -- Main Program --
hero1 = Hero("Layla",  100, 15)
hero2 = Hero("Zilong", 120, 20)

hero1.info()
hero2.info()
````

### Tugas Analisis 1
**Pertanyaan:** Apa yang terjadi jika kamu mengubah `hero1.hp` menjadi 500 setelah baris `hero1 = Hero(...)` ? Coba lakukan `print(hero1.hp)`.

````python
# Uji coba Tugas Analisis 1
hero1 = Hero("Layla", 100, 15)
print("HP awal :", hero1.hp)   # Output: 100

hero1.hp = 500                  # Mengubah atribut secara langsung
print("HP setelah diubah:", hero1.hp)  # Output: 500
hero1.info()
````

**Jawaban Analisis 1:**

Atribut `hp` milik `hero1` berhasil diubah menjadi `500`. Ini membuktikan bahwa **atribut publik** di Python bisa diubah dari mana saja, kapan saja. Itulah alasan **Encapsulation (Latihan 4)** diperlukan — supaya data seperti HP tidak bisa diutak-atik sembarangan dari luar class.

## Latihan 2 — Interaksi Antar Objek (Method)
**Tujuan:** Membuat dua objek saling berinteraksi — dalam hal ini, saling menyerang.

````python
class Hero:
    def __init__(self, name, hp, attack_power):
        self.name         = name
        self.hp           = hp
        self.attack_power = attack_power

    def info(self):
        print(f"Hero: {self.name} | HP: {self.hp} | Power: {self.attack_power}")

    # Method menyerang: Objek ini (self) menyerang objek lain (lawan)
    def serang(self, lawan):
        print(f"{self.name} menyerang {lawan.name}!")
        lawan.diserang(self.attack_power)

    # Method diserang: Menerima damage
    def diserang(self, damage):
        self.hp -= damage
        print(f"{self.name} terkena damage {damage}. Sisa HP: {self.hp}")

# -- Main Program --
hero1 = Hero("Layla",  100, 15)
hero2 = Hero("Zilong", 120, 20)

print("\n--- Pertarungan Dimulai ---")
hero1.serang(hero2)   # Layla menyerang Zilong
hero2.serang(hero1)   # Zilong membalas
````

### Tugas Analisis 2
**Pertanyaan:** Parameter `lawan` pada method `serang` menerima objek utuh, bukan sekadar string nama. Kenapa ini penting?

**Jawaban Analisis 2:**

Karena dengan menerima **objek utuh**, method `serang` bisa langsung:
1. **Mengakses atribut** lawan — misalnya `lawan.name` untuk nama dan `lawan.hp` untuk kondisi HP-nya.
2. **Memanggil method** milik lawan — yaitu `lawan.diserang(self.attack_power)`, sehingga pengurangan HP terjadi di dalam objek lawan itu sendiri. Data dikelola di tempat yang semestinya.

Kalau `lawan` cuma string nama, kita tidak bisa memanggil `lawan.diserang(...)` dan terpaksa mengelola HP dari luar — yang rawan error dan melanggar prinsip OOP.

## Latihan 3 — Pewarisan (Inheritance)
**Tujuan:** Membuat class `Mage` yang mewarisi sifat dasar `Hero` tapi punya kemampuan tambahan.

````python
class Hero:
    def __init__(self, name, hp, attack_power):
        self.name         = name
        self.hp           = hp
        self.attack_power = attack_power

    def info(self):
        print(f"Hero: {self.name} | HP: {self.hp} | Power: {self.attack_power}")

    def serang(self, lawan):
        print(f"{self.name} menyerang {lawan.name}!")
        lawan.diserang(self.attack_power)

    def diserang(self, damage):
        self.hp -= damage
        print(f"{self.name} terkena damage {damage}. Sisa HP: {self.hp}")


# Class Mage adalah anak dari class Hero
class Mage(Hero):
    def __init__(self, name, hp, attack_power, mana):
        # Memanggil constructor milik Parent (Hero)
        super().__init__(name, hp, attack_power)
        self.mana = mana

    def info(self):
        print(f"{self.name} [Mage] | HP: {self.hp} | Mana: {self.mana}")

    # Mage punya skill khusus
    def skill_fireball(self, lawan):
        if self.mana >= 20:
            print(f"{self.name} menggunakan Fireball ke {lawan.name}!")
            self.mana -= 20
            lawan.diserang(self.attack_power * 2)  # Damage 2x lipat
        else:
            print(f"{self.name} gagal skill! Mana tidak cukup.")


# -- Main Program Baru --
print("\n--- Update Class Hero ---")
eudora  = Mage("Eudora",  80, 30, 100)
balmond = Hero("Balmond", 200, 10)

eudora.info()
eudora.serang(balmond)          # Serangan biasa (warisan dari Hero)
eudora.skill_fireball(balmond)  # Skill khusus Mage
````

### Eksperimen: Hapus `super().__init__()` — lihat apa yang terjadi

````python
# Eksperimen Tugas Analisis 3: super() dihapus
class MageTanpaSuper(Hero):
    def __init__(self, name, hp, attack_power, mana):
        # super().__init__(name, hp, attack_power)  <-- sengaja dikomentari
        self.mana = mana

    def info(self):
        print(f"{self.name} [Mage] | HP: {self.hp} | Mana: {self.mana}")

try:
    eudora_broken = MageTanpaSuper("Eudora", 80, 30, 100)
    eudora_broken.info()   # Ini akan memicu AttributeError!
except AttributeError as e:
    print(f"[ERROR] {e}")
````

### Tugas Analisis 3

**Error yang muncul:**
```
AttributeError: 'MageTanpaSuper' object has no attribute 'name'
```

**Jawaban Analisis 3:**

Kalau `super().__init__(...)` dihapus, constructor milik class `Hero` tidak pernah dijalankan. Akibatnya, atribut `self.name`, `self.hp`, dan `self.attack_power` yang seharusnya dibuat oleh Hero tidak pernah terbentuk di objek `Mage`.

Nilainya sudah dikirim saat membuat objek, tapi tidak pernah disimpan ke `self.name` karena baris `super().__init__()` yang bertugas melakukan itu sengaja dihapus.

**Peran `super()`:** Fungsi ini adalah cara class Anak mendelegasikan inisialisasi atribut warisan ke class Induk, tanpa perlu menduplikasi kode yang sama.

## Latihan 4 — Enkapsulasi (Mengamankan Data HP)
**Tujuan:** Mencegah HP diubah sembarangan — misalnya jadi negatif atau di-cheat jadi 9999.

````python
class Hero:
    def __init__(self, nama, hp_awal):
        self.nama  = nama
        # Enkapsulasi: HP bersifat Private
        self.__hp  = hp_awal

    # GETTER: Cara resmi melihat HP
    def get_hp(self):
        return self.__hp

    # SETTER: Cara resmi mengubah HP (dengan validasi)
    def set_hp(self, nilai_baru):
        if nilai_baru < 0:
            self.__hp = 0          # HP tidak boleh negatif
        elif nilai_baru > 1000:
            print("Cheat terdeteksi! HP dimaksimalkan ke 1000 saja.")
            self.__hp = 1000
        else:
            self.__hp = nilai_baru

    def diserang(self, damage):
        sisa_hp = self.get_hp() - damage
        self.set_hp(sisa_hp)
        print(f"{self.nama} terkena damage {damage}. Sisa HP: {self.get_hp()}")


# -- Uji Coba --
hero1 = Hero("Layla", 100)

hero1.set_hp(-50)              # Coba set negatif
print("HP setelah set -50  :", hero1.get_hp())   # Output: 0

hero1.set_hp(9999)             # Coba cheat
print("HP setelah set 9999 :", hero1.get_hp())   # Output: 1000

hero1.set_hp(100)              # Reset normal
hero1.diserang(30)
print("HP setelah diserang :", hero1.get_hp())
````

### Eksperimen Tugas Analisis 4 — Name Mangling & Validasi

````python
# Eksperimen 1: Name Mangling
hero1 = Hero("Layla", 100)
print(f"Mencoba akses paksa: {hero1._Hero__hp}")
````

````python
# Eksperimen 2: Setter tanpa validasi
class HeroTanpaValidasi:
    def __init__(self, nama, hp_awal):
        self.nama = nama
        self.__hp = hp_awal

    def get_hp(self):
        return self.__hp

    def set_hp(self, nilai_baru):
        self.__hp = nilai_baru   # Tanpa validasi sama sekali

h = HeroTanpaValidasi("Zilong", 100)
h.set_hp(-100)
print("HP setelah set_hp(-100) tanpa validasi:", h.get_hp())   # Output: -100 (berbahaya!)
````

### Tugas Analisis 4

**Eksperimen 1 — Name Mangling:**

Nilai HP **tetap muncul**, tidak Error. Python menerapkan *Name Mangling*: atribut `__hp` pada class `Hero` secara internal disimpan sebagai `_Hero__hp`. Artinya masih bisa diakses kalau kita tahu caranya.

Tapi dalam praktik pemrograman yang baik, mengakses `_Hero__hp` dari luar class tetap dianggap salah. Ini soal konvensi, bukan larangan teknis. Kalau nama class diubah, akses paksa itu langsung Error — kode jadi rapuh dan susah dimaintain.

**Eksperimen 2 — Setter tanpa validasi:**

HP Hero bisa jadi **-100**, yang tidak masuk akal dalam game. Karakter seharusnya sudah mati di HP 0. Tanpa validasi, data bisa rusak diam-diam tanpa ada peringatan. Itulah fungsi Setter: ia menjadi **penjaga** yang memastikan nilai yang masuk selalu valid sebelum disimpan.

## Latihan 5 — Abstraction & Interface (Kontrak/Standar)
**Tujuan:** Memastikan semua class karakter wajib memiliki method `serang` dan `info`.

````python
from abc import ABC, abstractmethod

# 1. Interface / Abstract Class — KONTRAK
class GameUnit(ABC):

    @abstractmethod
    def serang(self, target):
        pass

    @abstractmethod
    def info(self):
        pass


# 2. Implementasi pada Class Konkret
class Hero(GameUnit):
    def __init__(self, nama):
        self.nama = nama

    def serang(self, target):
        print(f"Hero {self.nama} menebas {target}!")

    def info(self):
        print(f"Saya adalah Hero: {self.nama}")


class Monster(GameUnit):
    def __init__(self, jenis):
        self.jenis = jenis

    def serang(self, target):
        print(f"Monster {self.jenis} menggigit {target}!")

    def info(self):
        print(f"Saya adalah Monster: {self.jenis}")


# -- Uji Coba --
h = Hero("Alucard")
m = Monster("Serigala")

h.info()
m.info()
h.serang("Serigala")
m.serang("Alucard")
````

### Eksperimen Tugas Analisis 5 — Melanggar Kontrak & Instansiasi Abstract Class

````python
# Eksperimen 1: Melanggar kontrak — method serang dihapus dari Hero
from abc import ABC, abstractmethod

class GameUnit(ABC):
    @abstractmethod
    def serang(self, target): pass

    @abstractmethod
    def info(self): pass

class HeroBroken(GameUnit):
    def __init__(self, nama):
        self.nama = nama
    # method serang SENGAJA dihapus!
    def info(self):
        print(f"Saya Hero: {self.nama}")

try:
    h = HeroBroken("Alucard")   # Ini akan Error!
except TypeError as e:
    print(f"[ERROR] {e}")
````

````python
# Eksperimen 2: Mencoba instansiasi Abstract Class langsung
try:
    unit = GameUnit()   # Ini akan Error!
except TypeError as e:
    print(f"[ERROR] {e}")
````

### Tugas Analisis 5

**Eksperimen 1 — Melanggar Kontrak:**

Error yang muncul:
```
TypeError: Can't instantiate abstract class HeroBroken with abstract method serang
```

Python menolak membuat objek dari `HeroBroken` karena class itu belum memenuhi semua kewajiban dari `GameUnit`. Method `serang` yang ditandai `@abstractmethod` **wajib diimplementasikan** di setiap class turunan. Kalau lupa, program langsung error sebelum jalan — ini justru bagus karena mencegah bug yang tersembunyi.

**Eksperimen 2 — Instansiasi Abstract Class:**

Class `GameUnit` tidak bisa dijadikan objek karena ia hanya blueprint, bukan implementasi nyata. Ia tidak tahu *bagaimana* caranya menyerang — ia hanya menetapkan bahwa setiap unit game *harus bisa* menyerang. Fungsinya sebagai **kontrak** bagi semua class turunan, memastikan semuanya punya antarmuka yang seragam sehingga bisa diperlakukan secara konsisten.

## Latihan 6 — Polymorphism (Fleksibilitas Interaksi)
**Tujuan:** Menjalankan perintah yang sama (`serang`) ke berbagai jenis objek yang berbeda.

````python
# Parent Class
class Hero:
    def __init__(self, nama):
        self.nama = nama

    def serang(self):
        print("Hero menyerang dengan tangan kosong.")


# Child Class 1
class Mage(Hero):
    def serang(self):
        print(f"{self.nama} (Mage) menembakkan Bola Api! Boom!")


# Child Class 2
class Archer(Hero):
    def serang(self):
        print(f"{self.nama} (Archer) memanah dari jauh! Jleb!")


# Child Class 3
class Fighter(Hero):
    def serang(self):
        print(f"{self.nama} (Fighter) memukul dengan pedang! Slash!")


# -- Penerapan Polymorphism --
pasukan = [
    Mage("Eudora"),
    Archer("Miya"),
    Fighter("Zilong"),
    Mage("Gord")
]

print("--- PERANG DIMULAI ---")
for pahlawan in pasukan:
    pahlawan.serang()
````

### Eksperimen Tugas Analisis 6.1 — Tambah Class Healer (uji skalabilitas)

````python
# Tugas Analisis 6.1: Tambah class Healer TANPA mengubah kode loop
class Healer(Hero):
    def serang(self):
        print(f"{self.nama} tidak menyerang, tapi menyembuhkan teman!")


# Cukup tambahkan objek Healer ke list — loop tidak berubah sama sekali!
pasukan_baru = [
    Mage("Eudora"),
    Archer("Miya"),
    Fighter("Zilong"),
    Healer("Rafaela"),   # <-- ditambahkan
    Mage("Gord")
]

print("--- PERANG DIMULAI (dengan Healer) ---")
for pahlawan in pasukan_baru:
    pahlawan.serang()
````

### Eksperimen Tugas Analisis 6.2 — Ganti nama method jadi `tembak_panah`

````python
# Tugas Analisis 6.2: Archer menggunakan nama method berbeda
class ArcherBroken(Hero):
    def tembak_panah(self):   # Nama berbeda dari 'serang'!
        print(f"{self.nama} (Archer) memanah dari jauh! Jleb!")

pasukan_test = [
    Mage("Eudora"),
    ArcherBroken("Miya"),    # Archer yang rusak
    Fighter("Zilong")
]

print("--- Uji Nama Method Berbeda ---")
for pahlawan in pasukan_test:
    pahlawan.serang()   # ArcherBroken tidak punya 'serang' -> jatuh ke Hero default
````

### Tugas Analisis 6

**Analisis 6.1 — Uji Skalabilitas:**

Program berjalan lancar tanpa mengubah satu baris pun di kode loop. Ini kekuatan utama Polymorphism: cukup buat class baru yang mewarisi `Hero` dan implementasikan method `serang`, lalu masukkan objeknya ke list. Kode yang sudah ada tidak perlu disentuh — prinsip ini disebut **Open/Closed Principle**: terbuka untuk penambahan, tertutup untuk perubahan.

**Analisis 6.2 — Konsistensi Penamaan:**

Saat method di `ArcherBroken` dinamai `tembak_panah`, perintah `pahlawan.serang()` di dalam loop tidak memanggil `tembak_panah` — melainkan jatuh ke method `serang` warisan dari class `Hero`. Perilaku Archer pun salah, tapi tidak ada Error yang muncul. Ini jenis bug yang susah dilacak.

Dalam Polymorphism, **nama method harus sama** antara Parent dan semua Child-nya karena loop memanggil satu nama yang sama untuk semua objek. Konsistensi nama inilah yang memungkinkan satu perintah bekerja benar untuk semua jenis hero.
